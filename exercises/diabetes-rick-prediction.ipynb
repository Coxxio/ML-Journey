```python
import pandas as pd
pd.plotting.register_matplotlib_converters()
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

## Step 1: Load the data 

```python
filepath = '../data/diabetes_risk.csv'
diabetes_data = pd.read_csv(filepath)
diabetes_data.head()
```

```python
female_data = diabetes_data[diabetes_data['gender'] == 'Female']['diabetes_risk'].value_counts()
male_data = diabetes_data[diabetes_data['gender'] == 'Male']['diabetes_risk'].value_counts()

risk_order = ['Low', 'Moderate', 'High']
x = list(range(len(risk_order)))

female_vals = female_data.reindex(risk_order).values
male_vals = male_data.reindex(risk_order).values

sns.barplot(x=[i - 0.2 for i in x], y=female_vals, alpha=0.8, color='blue', label='Female')
sns.barplot(x=[i + 0.2 for i in x], y=male_vals, alpha=0.8, color='orange', label='Male')

plt.xticks(x, risk_order)
plt.xlabel('Diabetes risk')
plt.ylabel('Count')
plt.title('Diabetes risk by gender')
plt.legend()
```
