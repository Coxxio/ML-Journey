# ================================
# 1. Vectors — What even are they?
# ================================

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# A verctor its just a list of number with direction and magnitude
v1 = np.array([1, 2])
v2 = np.array([3, -1])
```

```python
# Visualize it
fig, ax = plt.subplots()
ax.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='steelblue')
ax.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='orange')
ax.set_xlim(-2, 5); ax.set_ylim(-2, 5)
ax.grid(True)
plt.title("Vectors [2, 1] and [3, -1]")
plt.show()
```

```python
# ==============================================================================
# Vector addition
# To add two vectors, we just add their corresponding components together
# For example, to add v1 and v2, we would do:
# v1 + v2 = v1[0] + v2[0], v1[1] + v2[1]
# ==============================================================================
v3 = v1 + v2

fig, ax = plt.subplots()
ax.quiver(v2[0], v2[1], v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='steelblue', label='v1')
ax.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='orange', label='v2')
ax.quiver(0, 0, v3[0], v3[1], angles='xy', scale_units='xy', scale=1, color='green', label='v3')
ax.set_xlim(-2, 5); ax.set_ylim(-2, 5)
ax.grid(True)
plt.title("v1 + v2")
plt.legend()
plt.show()
```

# ================================
# 2. Linear combination and span
# ================================

```python
v4 = np.array([2, 1])
v5 = np.array([4, 2])

fig2, ax2 = plt.subplots()
ax2.quiver(0, 0, v5[0], v5[1], angles='xy', scale_units='xy', scale=1, color='orange', label='v5')
ax2.quiver(0, 0, v4[0], v4[1], angles='xy', scale_units='xy', scale=1, color='steelblue', label='v4')
ax2.set_xlim(-2, 5); ax2.set_ylim(-2, 5)
ax2.grid(True)
plt.title("v1 + v2")
plt.legend()
plt.show()
```

# ================================
# 3. Linear transformations
# ================================

```python
v6 = np.array([3, -2])
v7 = np.array([2, 1])
v8a = np.array([5, 7])
transformation = [((v6[0] * v8a[0])+ (v7[0] * v8a[1])), ((v6[1] * v8a[0])+ (v7[1] * v8a[1]))]
v8b = np.array(transformation)

fig3a, ax3a = plt.subplots()
ax3a.quiver(0, 0, 0, 1, angles='xy', scale_units='xy', scale=1, color='orange', label='v6')
ax3a.quiver(0, 0, 1, 0, angles='xy', scale_units='xy', scale=1, color='green', label='v7')
ax3a.quiver(0, 0, v8a[0], v8a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='v8')
ax3a.set_xlim(-2, 8); ax3a.set_ylim(-2, 8)
ax3a.grid(True)
plt.title("Untransformed")
plt.legend()
plt.show()

fig3b, ax3b = plt.subplots()
ax3b.quiver(0, 0, v6[0], v6[1], angles='xy', scale_units='xy', scale=1, color='orange', label='v6')
ax3b.quiver(0, 0, v7[0], v7[1], angles='xy', scale_units='xy', scale=1, color='green', label='v7')
ax3b.quiver(0, 0, v8b[0], v8b[1], angles='xy', scale_units='xy', scale=1, color='blue', label='v8')
ax3b.set_xlim(-2, 29); ax3b.set_ylim(-20, 10)
ax3b.grid(True)
plt.title("Transformed")
plt.legend()
plt.show()
```
